```python
import ollama

from config import MODEL_NAME
from prompts import PROMPT_A, PROMPT_B, PROMPT_C
```

```python
def run_prompt_experiment(system_prompt, question):
    response = ollama.chat(
        model=MODEL_NAME,
        messages=[
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": question
            }
        ]
    )

    return response["message"]["content"]
```

```python
questions = [
    "Explain REST APIs.",
    "Explain recursion.",
    "What is dependency injection?Show me an example."
]
```

```python
for question in questions:
    print(f"\nQuestion: {question}")
    print(run_prompt_experiment(PROMPT_A, question))
```

```python
for question in questions:
    print(f"\nQuestion: {question}")
    print(run_prompt_experiment(PROMPT_B, question))
```

```python
for question in questions:
    print(f"\nQuestion: {question}")
    print(run_prompt_experiment(PROMPT_C, question))
```
